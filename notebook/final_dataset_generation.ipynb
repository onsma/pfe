# Generation de `final_dataset`

Ce notebook genere un dataset synthetique de 10 000 clients PM en reutilisant les scores BFPME historiques et des regles de realisme pour les donnees client, credit, financieres, comportementales, macroeconomiques et la cible risque.

```python
from pathlib import Path
import runpy
import pandas as pd

project_root = Path(r"d:/BFPME")
script_path = project_root / "scripts" / "generate_final_dataset.py"
output_path = project_root / "data" / "final_dataset.csv"

print(f"Script: {script_path}")
print(f"Output: {output_path}")
```

```python
runpy.run_path(str(script_path), run_name="__main__")
```

```python
df = pd.read_csv(output_path)
print(df.shape)
df.head()
```

```python
key_cols = [
    "score_bfpme_global",
    "score_global_interne",
    "retards_paiement_jours_moyen",
    "montant_impayes",
    "taux_inflation",
    "taux_directeur",
    "croissance_pib",
    "taux_chomage",
    "pd_target",
    "default_flag",
]

print(df[key_cols].describe(include="all").T)
```
